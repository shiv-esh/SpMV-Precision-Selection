```python
# Read the CSV files
df1 = pd.read_csv('output_15k.csv')  # Replace with your first file name
df2 = pd.read_csv('spmv_15k.csv')  
```

```python
df1['matrix'] = df1['matrix'].str.replace(r'\.mtx$', '', regex=True)
```

```python
# Clean up the 'matrix' column by stripping spaces and converting to lowercase
df1['matrix'] = df1['matrix'].str.strip().str.lower()
df2['matrix'] = df2['matrix'].str.strip().str.lower()
```
